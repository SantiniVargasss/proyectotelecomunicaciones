```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
# 1.2 Parámetros de la señal
fs = 10000       # Frecuencia de muestreo (Hz)
fc = 1000        # Frecuencia portadora (Hz)
Rs = 100         # Tasa de símbolos

# 1.3 Funciones generadoras
def generar_ask(bits, fs, fc, Rs):
    t = np.linspace(0, len(bits) / Rs, len(bits) * int(fs / Rs), endpoint=False)
    portadora = np.cos(2 * np.pi * fc * t)
    simbolos = np.repeat(bits, int(fs / Rs))
    return simbolos * portadora

def generar_fsk(bits, fs, fc1, fc2, Rs):
    t = np.linspace(0, len(bits) / Rs, len(bits) * int(fs / Rs), endpoint=False)
    frecuencias = np.repeat(bits, int(fs / Rs))
    f_inst = np.where(frecuencias == 1, fc2, fc1)
    return np.cos(2 * np.pi * f_inst * t)

def generar_psk(bits, fs, fc, Rs):
    t = np.linspace(0, len(bits) / Rs, len(bits) * int(fs / Rs), endpoint=False)
    fases = np.repeat(bits, int(fs / Rs)) * np.pi
    return np.cos(2 * np.pi * fc * t + fases)

def generar_qam16(num_simbolos, fs, fc, Rs):
    t = np.linspace(0, num_simbolos / Rs, num_simbolos * int(fs / Rs), endpoint=False)
    i_data = np.random.choice([-3, -1, 1, 3], num_simbolos)
    q_data = np.random.choice([-3, -1, 1, 3], num_simbolos)
    i_signal = np.repeat(i_data, int(fs / Rs)) * np.cos(2 * np.pi * fc * t)
    q_signal = np.repeat(q_data, int(fs / Rs)) * np.sin(2 * np.pi * fc * t)
    return i_signal + q_signal
```

```python
# 1.4 Generar el conjunto de datos (1,000 muestras por cada modulación)
X = []
y = []
muestras_por_clase = 1000

print("Generando dataset, por favor espera un momento...")
for _ in range(muestras_por_clase):
    bits_aleatorios = np.random.randint(0, 2, 10)

    # ASK
    X.append(generar_ask(bits_aleatorios, fs, fc, Rs))
    y.append('ASK')

    # FSK (usando fc y fc*2 como portadoras)
    X.append(generar_fsk(bits_aleatorios, fs, fc, fc * 2, Rs))
    y.append('FSK')

    # PSK
    X.append(generar_psk(bits_aleatorios, fs, fc, Rs))
    y.append('PSK')

    # QAM (16-QAM con 10 símbolos)
    X.append(generar_qam16(10, fs, fc, Rs))
    y.append('QAM')

X = np.array(X)
y = np.array(y)

# 1.5 Añadir Ruido Gaussiano Blanco Aditivo (AWGN)
def agregar_awgn(senal, snr_db):
    potencia_senal = np.mean(senal ** 2)
    snr_lineal = 10 ** (snr_db / 10)
    potencia_ruido = potencia_senal / snr_lineal
    ruido = np.random.normal(0, np.sqrt(potencia_ruido), len(senal))
    return senal + ruido

print("Agregando ruido a las señales...")
X_ruidoso = np.array([agregar_awgn(sig, np.random.uniform(5, 30)) for sig in X])

# 1.6 Separación de datos (Entrenamiento 80% y Prueba 20%)
X_train, X_test, y_train, y_test = train_test_split(
    X_ruidoso, y, test_size=0.20, random_state=42, stratify=y
)

print("¡Listo! Datos generados y separados con éxito.")
print(f"Dimensiones de X_train (Entrenamiento): {X_train.shape}")
print(f"Dimensiones de X_test (Prueba): {X_test.shape}")
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt

print("--- 1. PREPARACIÓN Y CODIFICACIÓN DE DATOS ---")
# Añadir la dimensión de canales (1 canal) requerida por Conv1D
X_train_cnn = X_train[..., np.newaxis]  # Forma: (3200, 1000, 1)
X_test_cnn = X_test[..., np.newaxis]    # Forma: (800, 1000, 1)

# Convertir las etiquetas de texto ('ASK', 'FSK', etc.) a números enteros (0, 1, 2, 3)
encoder = LabelEncoder()
y_train_encoded = encoder.fit_transform(y_train)
y_test_encoded = encoder.transform(y_test)

print("Nuevas dimensiones de X_train:", X_train_cnn.shape)
print("Etiquetas codificadas correctamente:", np.unique(y_train_encoded))

print("\n--- 2. DISEÑO DEL MODELO CNN 1D ---")
input_shape = (X_train_cnn.shape[1], 1)

model = models.Sequential([
    layers.Conv1D(filters=32, kernel_size=3, activation='relu', input_shape=input_shape),
    layers.MaxPooling1D(pool_size=2),

    layers.Conv1D(filters=64, kernel_size=3, activation='relu'),
    layers.MaxPooling1D(pool_size=2),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(4, activation='softmax') # 4 neuronas para las 4 modulaciones
])

model.summary()

print("\n--- 3. COMPILACIÓN DEL MODELO ---")
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\n--- 4. ENTRENAMIENTO DEL MODELO ---")
EPOCHS = 15
BATCH_SIZE = 32

history = model.fit(
    X_train_cnn, y_train_encoded,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.2
)

print("\n--- 5. GUARDAR EL MODELO ---")
model.save('detector_modulaciones_cnn.keras')
print("¡Modelo entrenado y guardado exitosamente como 'detector_modulaciones_cnn.keras'!")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import tensorflow as tf
from tensorflow.keras import layers, models

# ==========================================
# 1. GENERACIÓN DE DATOS Y DIVISIÓN
# ==========================================
fs = 10000  # Frecuencia de muestreo (Hz)
fc = 1000  # Frecuencia portadora (Hz)
Rs = 100  # Tasa de símbolos


def generar_ask(bits, fs, fc, Rs):
  t = np.linspace(
      0, len(bits) / Rs, len(bits) * int(fs / Rs), endpoint=False
  )
  portadora = np.cos(2 * np.pi * fc * t)
  simbolos = np.repeat(bits, int(fs / Rs))
  return simbolos * portadora


def generar_fsk(bits, fs, fc1, fc2, Rs):
  t = np.linspace(
      0, len(bits) / Rs, len(bits) * int(fs / Rs), endpoint=False
  )
  frecuencias = np.repeat(bits, int(fs / Rs))
  f_inst = np.where(frecuencias == 1, fc2, fc1)
  return np.cos(2 * np.pi * f_inst * t)


def generar_psk(bits, fs, fc, Rs):
  t = np.linspace(
      0, len(bits) / Rs, len(bits) * int(fs / Rs), endpoint=False
  )
  fases = np.repeat(bits, int(fs / Rs)) * np.pi
  return np.cos(2 * np.pi * fc * t + fases)


def generar_qam16(num_simbolos, fs, fc, Rs):
  t = np.linspace(
      0, num_simbolos / Rs, num_simbolos * int(fs / Rs), endpoint=False
  )
  i_data = np.random.choice([-3, -1, 1, 3], num_simbolos)
  q_data = np.random.choice([-3, -1, 1, 3], num_simbolos)
  i_signal = np.repeat(i_data, int(fs / Rs)) * np.cos(2 * np.pi * fc * t)
  q_signal = np.repeat(q_data, int(fs / Rs)) * np.sin(2 * np.pi * fc * t)
  return i_signal + q_signal


def agregar_awgn_lote(senales, snr_db):
  """Añade AWGN a un arreglo 2D de forma (N, longitud) para un SNR en dB."""
  senales = np.asarray(senales, dtype=np.float64)
  potencia_senal = np.mean(senales**2, axis=1, keepdims=True)
  snr_lineal = 10 ** (snr_db / 10.0)
  potencia_ruido = potencia_senal / snr_lineal
  ruido = np.random.normal(0, np.sqrt(potencia_ruido), size=senales.shape)
  return senales + ruido


X = []
y = []
muestras_por_clase = 1000

print("Generando dataset...")
for _ in range(muestras_por_clase):
  bits_aleatorios = np.random.randint(0, 2, 10)

  X.append(generar_ask(bits_aleatorios, fs, fc, Rs))
  y.append("ASK")

  X.append(generar_fsk(bits_aleatorios, fs, fc, fc * 2, Rs))
  y.append("FSK")

  X.append(generar_psk(bits_aleatorios, fs, fc, Rs))
  y.append("PSK")

  X.append(generar_qam16(10, fs, fc, Rs))
  y.append("QAM")

X = np.array(X)
y = np.array(y)

print("Agregando ruido inicial (SNR aleatorio entre 5 y 30 dB)...")
X_ruidoso = np.array(
    [agregar_awgn_lote(sig[np.newaxis, :], np.random.uniform(5, 30))[0] for sig in X]
)

X_train, X_test, y_train, y_test = train_test_split(
    X_ruidoso, y, test_size=0.20, random_state=42, stratify=y
)

# Preparación para CNN 1D y etiquetas numéricas
X_train_cnn = X_train[..., np.newaxis]
X_test_cnn = X_test[..., np.newaxis]

encoder = LabelEncoder()
y_train_encoded = encoder.fit_transform(y_train)
y_test_encoded = encoder.transform(y_test)
clases = list(encoder.classes_)

print("X_train_cnn:", X_train_cnn.shape, "| X_test_cnn:", X_test_cnn.shape)
print("Clases:", clases)


# ==========================================
# 2. DISEÑO, ENTRENAMIENTO Y GUARDADO
# ==========================================
input_shape = (X_train_cnn.shape[1], 1)

model = models.Sequential([
    layers.Conv1D(
        filters=32, kernel_size=3, activation="relu", input_shape=input_shape
    ),
    layers.MaxPooling1D(pool_size=2),
    layers.Conv1D(filters=64, kernel_size=3, activation="relu"),
    layers.MaxPooling1D(pool_size=2),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(len(clases), activation="softmax"),
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

EPOCHS = 15
BATCH_SIZE = 32

history = model.fit(
    X_train_cnn,
    y_train_encoded,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    verbose=1,
)

nombre_modelo = "detector_modulaciones_cnn.keras"
model.save(nombre_modelo)
print(f"Modelo guardado como '{nombre_modelo}'")


# ==========================================
# 3. EVALUACIÓN: MATRIZ DE CONFUSIÓN Y SNR
# ==========================================
modelo_eval = tf.keras.models.load_model(nombre_modelo)

y_pred_probs = modelo_eval.predict(X_test_cnn, verbose=0)
y_pred = np.argmax(y_pred_probs, axis=1)
y_true = y_test_encoded

cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=clases,
    yticklabels=clases,
)
plt.title("Matriz de Confusión - Clasificación de Modulaciones")
plt.xlabel("Predicción del Modelo")
plt.ylabel("Modulación Real")
plt.tight_layout()
plt.show()

print(classification_report(y_true, y_pred, target_names=clases))

# Barrido de SNR sobre X_test (con dimensión de canal añadida)
niveles_snr = [-10, -5, 0, 5, 10, 15, 20, 25, 30]
precisiones = []

for snr in niveles_snr:
  X_snr = agregar_awgn_lote(X_test, snr)[..., np.newaxis]
  preds = np.argmax(modelo_eval.predict(X_snr, verbose=0), axis=1)
  acc = accuracy_score(y_true, preds)
  precisiones.append(acc * 100)
  print(f"SNR: {snr:3d} dB -> Precisión: {acc*100:.2f}%")

plt.figure(figsize=(8, 5))
plt.plot(niveles_snr, precisiones, marker="o", linewidth=2, color="#1f77b4")
plt.axhline(25, color="red", linestyle="--", label="Azar (4 clases = 25%)")
plt.title("Rendimiento del Modelo vs. Nivel de Ruido (SNR)")
plt.xlabel("SNR (dB) [Menor = Más Ruido | Mayor = Más Señal]")
plt.ylabel("Precisión (%)")
plt.grid(True, linestyle=":", alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```
